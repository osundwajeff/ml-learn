# Multiplication

```python
# vector - vector multiplication

print(f" {u} * {v}")
u * v
```

```python
# vector - vector multiplication (dot product)
# function

def vector_vector_multiplication(u, v):
    # assert vectors of same shape
    assert u.shape[0] == v.shape[0]

    # number of elements in u
    n = u.shape[0]

    # initialise result
    result = 0

    # loop through elements
    # as you sum it to the result
    for i in range(n):
        result = result + u[i] * v[i]

    # return result
    return result
```

```python
# run against our vectors u & v
# returns the value of our own custom dot product

vector_vector_multiplication(u, v)
```

```python
# compare previous result with numpy dot product

u.dot(v)
```

```python
# matrix - vector multiplication
# it's basically the dot product of each row of the matrix with the vector

def matrix_vector_multiplication(U, v):

    # asserts if matrix column size is the same as the column size of the vector
    assert U.shape[1] == v.shape[0]

    # gets the number of rows of the matrix
    num_rows = U.shape[0]

    # create a matrix with same number of rows filled of zeros
    result = np.zeros(num_rows)

    # loop through the matrix.rows
    for i in range(num_rows):
        # for each row we compute the dot matrix
        # hence use of this fucntion at this point
        # so each row of the matrix is treated as
        # a single vector, hence U[i]
        result[i] = vector_vector_multiplication(U[i], v)

    return result
```

```python
# test the function
U = np.array([
    [2, 4, 5, 6],
    [1, 2, 1, 2],
    [3, 1, 2, 1],
])

v = np.array([1, 0.5, 2, 1])

matrix_vector_multiplication(U, v)
```

```python
# compare with in-built dot product

U.dot(v)
```

## matrix - matrix multiplication

```python
# matrix - matrix multiplication

# matrix V
V = np.array([
    [1, 1, 2],
    [0, 0.5, 1],
    [0, 2, 1],
    [2, 1, 0]
])

V.shape
```

```python
# matrix - matrix multiplication function

def matrix_matrix_mutltiplication(U, V):

    # checks that we can multiply matrix U & V
    assert U.shape[1] == V.shape[0]

    # get num of rows and cols from the 2 matrixes
    num_rows = U.shape[0]
    num_cols = V.shape[1]

    # result will take shape from both matrixes
    result = np.zeros((num_rows, num_cols))

    # loop through num of cols from matrix V
    # Then for each col treat it as a vector
    # Then do a matrix - vector multiplication (U * V[:, i])
    for i in range(num_cols):
        vi = V[:, i]
        Uvi = matrix_vector_multiplication(U, vi)
        result[:, i] = Uvi
        # Simpler
        # result[:, i] = matrix_vector_multiplication(U, V[:, i])

    return result
```

```python
matrix_matrix_mutltiplication(U, V)
```

```python
U.dot(V)
```

# Inverse

```python
# Identity matrix comes in handy here (explaining what an inverse matrix is)
# When You multiply the inverse of a matrix by the matrix you get the identity matrix
# Note: only square matrices have an inverse, so we need to take a subset of V for this example
# Note: Vs is V subset

Vs = V[[0, 1, 2]]

Vs
```

```python
# computing inverse of Vs

Vs_inv = np.linalg.inv(Vs)

Vs_inv
```

```python
# multiply the inverse by the matrix so as to get the identity matrix

Vs_inv.dot(Vs)
```
